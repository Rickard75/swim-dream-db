```python
# Python Tool : read specific section of a file
from pathlib import Path

file_path = Path.cwd().parent / "DOCS" / ("test.txt")

def read_file(file_path, start_line, end_line):
    with open(file_path, "r", encoding='utf-8') as selected_file: # "r" means read mode
        lines = selected_file.readlines() 
    return lines[start_line-1:end_line]

result = read_file(file_path, 1, 4)

for line in result:
    print(line.strip())
```
