```python
# Python Tool : read specific section of a file
import os
def read_file(file_path, start_line, end_line):
    with open(file_path, "r") as selected_file: # "r" means read mode
        lines = selected_file.readlines() 
    return lines[start_line-1:end_line]

print(read_file(r"..\DOCS\test.txt", 1, 5))

print("... Current directory", os.getcwd())
print("First 10 lines of the file:")
```
